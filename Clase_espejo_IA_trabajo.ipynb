## Paso 1: preparar el entorno

Ejecuta las dos celdas siguientes. No tienes que cambiar nada en ellas.

La primera instala la librería que usamos para hablar con la IA (se demora unos segundos). La segunda crea la función `preguntar_ia`, que es la que vamos a usar todo el tiempo. Cuando termine, debe aparecer el mensaje *Todo listo*.

```python
!pip install -q openai
```

```python
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get("GROQ_API_KEY"),
                base_url="https://api.groq.com/openai/v1")
MODELO = "openai/gpt-oss-120b"

def preguntar_ia(mensaje, personalidad="Eres un asistente amable. Respondes en español y de forma breve."):
    # Le manda el mensaje a la IA y devuelve la respuesta como texto
    try:
        r = client.chat.completions.create(
            model=MODELO,
            messages=[{"role": "system", "content": personalidad},
                      {"role": "user", "content": mensaje}])
        return r.choices[0].message.content
    except Exception as e:
        if "429" in str(e):
            return "Se alcanzó el límite de uso. Espera un minuto y vuelve a intentar."
        return "Error: " + str(e)

print("Todo listo, ya puedes usar preguntar_ia()")
```

## Paso 3: un chatbot con `while`

Con un `while`, un `input` y un `if` ya tenemos un chatbot.

Ejecuta la celda. Abajo aparece un cuadro para escribir: escribe algo y presiona **Enter**. Cuando quieras terminar, escribe `salir`.

```python
while True:
    texto = input("Tú: ")
    if texto.lower() == "salir":
        print("Chao!")
        break
    print("IA:", preguntar_ia(texto))
    print()
```

## Nivel 3: un ejemplo

Este es un quiz. La IA inventa las preguntas y revisa las respuestas, pero el puntaje lo lleva el programa con una variable. Úsenlo como idea o modifíquenlo.

Al ejecutarlo, escriban un tema (por ejemplo *fútbol* o *historia de Colombia*) y respondan las 3 preguntas.

```python
def quiz(tema, num_preguntas=3):
    puntaje = 0
    for i in range(1, num_preguntas + 1):
        pregunta = preguntar_ia("Escribe UNA pregunta corta y fácil sobre " + tema + ". Solo la pregunta, sin la respuesta.")
        print("\nPregunta", i, ":", pregunta)
        resp = input("Tu respuesta: ")
        veredicto = preguntar_ia("Pregunta: " + pregunta + "\nRespuesta: " + resp +
                                 "\nSi la respuesta es correcta, empieza con la palabra CORRECTO. "
                                 "Si no, empieza con INCORRECTO. Luego explica en una frase.")
        print(veredicto)
        if veredicto.strip().upper().startswith("CORRECTO"):
            puntaje += 1
    print("\nPuntaje final:", puntaje, "de", num_preguntas)

tema = input("¿De qué tema quieres el quiz? ")
quiz(tema)
```
