## Paso 1: preparar el entorno

Ejecuta las dos celdas siguientes. No tienes que cambiar nada en ellas.

La primera instala la librería que usamos para hablar con la IA (se demora unos segundos). La segunda crea la función `preguntar_ia`, que es la que vamos a usar todo el tiempo. Cuando termine, debe aparecer el mensaje *Todo listo*.

````python
!pip install -q openai
````

````python
from openai import OpenAI
from google.colab import userdata

client = OpenAI(api_key=userdata.get("GROQ_API_KEY"),
                base_url="https://api.groq.com/openai/v1")
MODELO = "openai/gpt-oss-120b"

def preguntar_ia(mensaje, personalidad="Eres un asistente amable. Respondes en español y de forma breve."):
    # Le manda el mensaje a la IA y devuelve la respuesta como texto
    try:
        r = client.chat.completions.create(
            model=MODELO,
            messages=[{"role": "system", "content": personalidad},
                      {"role": "user", "content": mensaje}])
        return r.choices[0].message.content
    except Exception as e:
        if "429" in str(e):
            return "Se alcanzó el límite de uso. Espera un minuto y vuelve a intentar."
        return "Error: " + str(e)

print("Todo listo, ya puedes usar preguntar_ia()")
````

## Paso 3: un chatbot con `while`

Con un `while`, un `input` y un `if` ya tenemos un chatbot.

Ejecuta la celda. Abajo aparece un cuadro para escribir: escribe algo y presiona **Enter**. Cuando quieras terminar, escribe `salir`.

## Paso 5: IA y datos

Ahora vamos a juntar la IA con análisis de datos. La idea es:

1. Pedirle a la IA que invente unos datos (como si fueran de un negocio real).
2. Analizarlos con **pandas**, la librería de Python para trabajar con tablas.
3. Pedirle a la IA que nos ayude a interpretar lo que encontramos.

Ejecuta la celda. Crea la función `generar_datos` y con ella pide 30 ventas inventadas de una tienda de barrio. Al final se muestran las primeras filas de la tabla.

````python
import pandas as pd
import io

def generar_datos(descripcion, filas=20):
    # Le pide a la IA un CSV con datos inventados y lo convierte en una tabla de pandas
    pedido = ("Genera un archivo CSV con " + str(filas) + " filas de datos ficticios sobre: " + descripcion + ". "
              "La primera línea debe tener los nombres de las columnas, separados por comas. "
              "No uses comas dentro de los valores. "
              "Responde SOLO con el CSV, sin explicaciones y sin ```.")
    texto = preguntar_ia(pedido)
    texto = texto.replace("```csv", "").replace("```", "").strip()
    return pd.read_csv(io.StringIO(texto))

ventas = generar_datos("ventas de una tienda de barrio en Bogotá, con las columnas producto, categoria, "
                       "cantidad, precio_unitario (en pesos, número entero) y dia_semana", 30)
ventas.head()
````

Con pandas podemos calcular cosas sobre la tabla. Esta celda crea una columna con el total de cada venta, suma las ventas por categoría y hace una gráfica por día de la semana.

Si sale un error de que una columna no existe, es porque la IA les puso otros nombres. Revísalos con `ventas.columns` y ajusta los nombres en la celda (o vuelve a generar los datos).

````python
ventas["total"] = ventas["cantidad"] * ventas["precio_unitario"]

print("Ventas totales: $", ventas["total"].sum())
print()

por_categoria = ventas.groupby("categoria")["total"].sum().sort_values(ascending=False)
print(por_categoria)

ventas.groupby("dia_semana")["total"].sum().plot(kind="bar", title="Ventas por día de la semana");
````

Por último, le pasamos a la IA el resumen que calculamos y le pedimos que lo interprete. Fíjense que **los cálculos los hace pandas** y la IA solo ayuda a explicarlos: así es más confiable que pedirle a la IA que haga las cuentas.

````python
analista = "Eres un analista de datos. Respondes en español, de forma clara y corta."
pregunta = ("Estas son las ventas totales por categoría de una tienda de barrio:\n" + por_categoria.to_string() +
            "\n\nDa tres conclusiones cortas y una recomendación para el dueño de la tienda.")
print(preguntar_ia(pregunta, analista))
````

Si quieren guardar los datos para usarlos después, pueden descargarlos como archivo:

````python
ventas.to_csv("ventas.csv", index=False)
print("Archivo guardado. Lo encuentras en el ícono de la carpeta, en la barra de la izquierda.")
````

## Nivel 3: un ejemplo

Este es un quiz. La IA inventa las preguntas y revisa las respuestas, pero el puntaje lo lleva el programa con una variable. Úsenlo como idea o modifíquenlo.

Al ejecutarlo, escriban un tema (por ejemplo *fútbol* o *historia de Colombia*) y respondan las 3 preguntas.

````python
def quiz(tema, num_preguntas=3):
    puntaje = 0
    for i in range(1, num_preguntas + 1):
        pregunta = preguntar_ia("Escribe UNA pregunta corta y fácil sobre " + tema + ". Solo la pregunta, sin la respuesta.")
        print("\nPregunta", i, ":", pregunta)
        resp = input("Tu respuesta: ")
        veredicto = preguntar_ia("Pregunta: " + pregunta + "\nRespuesta: " + resp +
                                 "\nSi la respuesta es correcta, empieza con la palabra CORRECTO. "
                                 "Si no, empieza con INCORRECTO. Luego explica en una frase.")
        print(veredicto)
        if veredicto.strip().upper().startswith("CORRECTO"):
            puntaje += 1
    print("\nPuntaje final:", puntaje, "de", num_preguntas)

tema = input("¿De qué tema quieres el quiz? ")
quiz(tema)
````
