```python
from openai import OpenAI
from google.colab import userdata

try:
    client = OpenAI(api_key=userdata.get("GROQ_API_KEY"),
                    base_url="https://api.groq.com/openai/v1")
    r = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user",
                   "content": "Saluda en una sola frase a unos estudiantes de programación en Colombia."}])
    print("✅ ¡Funciona! La IA respondió:\n")
    print(r.choices[0].message.content)
except Exception as e:
    print("❌ Algo falló. Revisa que el secreto se llame GROQ_API_KEY y tenga activado el acceso al notebook.")
    print("Detalle del error:", e)
```
